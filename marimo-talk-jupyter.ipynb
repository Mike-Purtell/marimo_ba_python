### Hello World of dataframes
Multiplication tables are often the first structured table people learn

```python
df_pandas = (
    pd.DataFrame(
        {str(i): [j * i for j in range(1, 10)] for i in range(1,10)}, 
        dtype='uint8'
    )
)
df_pandas # displays columns, index. No data types or dataframe shape
```

### Dataframe displays in Jupyter
The Pandas DataFrame above doesn’t display its datatypes or shape. The Polars DataFrame below includes both. Each DataFrame is simply a different representation of the same experimental data.


```python
df_polars = pl.from_pandas(df_pandas)
df_polars  # no index with polars. display has columns,  data types or shape
```

### Bring your own data viz
Jupyter does not provide built-in data visualization tools. 
Libraries like Plotly, Matplotlib, Altair or others are required.

```python

x_values = df_polars['1'].to_list()
fig = go.Figure([
        go.Scatter(
        x=x_values,
        y=df_polars[c].to_list(),
        mode='lines',
        name=c + '*X')
    for c in df_polars.columns
])
fig.update_layout(
    title='3rd Grade Times Table - 0s included for STEM students',
    xaxis_title='Factor',
    yaxis_title='Product',
    template='simple_white',
    hovermode='x unified',
    showlegend=False,
    width=(w:=600),
    height=w,
    margin=dict(r=90))
fig.update_xaxes(
    range=[min(x_values)-0.5, max(x_values)+0.5],
    tickmode='array',
    tickvals=x_values,
    ticktext=[str(value) for value in x_values],
    )
fig.update_yaxes(
    tickmode='array',
    tickvals=list(range(0, 101, 10)),
    showgrid=True,
    gridcolor='lightgray',
    griddash='dash')
fig.update_traces(cliponaxis=False)

for column in df_polars.columns:
    fig.add_annotation(
        x=x_values[-1],
        y=df_polars[column].to_list()[-1],
        text=column + ' X',
        showarrow=False,
        xanchor='left',
        xshift=10,
        yanchor='middle')

fig.show()
```

```python
sample_size = 2**16
print(f'{sample_size = }')
df = (
    pl.DataFrame({
        'X': [(360 * i) / sample_size for i in range(sample_size)],
        'SINE': [math.sin((2* math.pi * i) / sample_size) for i in range(sample_size)],
        'COS': [math.cos((2* math.pi * i) / sample_size) for i in range(sample_size)]
    })
)
df
```

```python
fig = px.line(
    df,
    'X',
    ['SINE', 'COS'],
    title='Sine and Cosine Functions',
    template='simple_white',
    width=800, 
    labels={
        'X': 'Degrees',
        'value': 'Function Value',
        'variable': ''
    },
    line_shape='hvh'
) 

fig.update_yaxes(
    range=[-1.1, 1.1],
    tickmode='array',
    tickvals=[
        -1, 
        -math.sqrt(3)/2, 
        -math.sqrt(2)/2, 
        -0.5,
        0, 
        0.5, 
        math.sqrt(2)/2, 
        math.sqrt(3)/2, 
        1
    ],
    ticktext=[
        '-1', 
        '-√3/2', 
        '-√2/2',
        '-0.5',
        '0', 
        '0.5', 
        '√2/2',
        '√3/2', 
        '1'
    ],
    showgrid=False,
)

fig.update_xaxes(
    range=[0, 360],
    tickmode='array',
    tickvals=list(range(0, 361, 30)),
    ticktext=[str(value) for value in range(0, 361, 30)],
    showgrid=True,
)

hline_args = dict(
    line_width=2,
    line_dash='solid', 
    line_color='#808080'
)
fig.add_hline(y=-1, **hline_args)
fig.add_hline(y=-math.sqrt(3)/2, **hline_args)
fig.add_hline(y=-math.sqrt(2)/2, **hline_args)
fig.add_hline(y=-0.5, **hline_args)
fig.add_hline(y=0, **hline_args)
fig.add_hline(y=0.5, **hline_args)
fig = px.line(
    df,
    'X',
    ['SINE', 'COS'],
    title='Sine and Cosine Functions',
    template='simple_white',
    width=800, 
    labels={
        'X': 'Degrees',
        'value': 'Function Value',
        'variable': ''
    },
    line_shape='hvh'
) 

fig.update_yaxes(
    range=[-1.1, 1.1],
    tickmode='array',
    tickvals=[
        -1, 
        -math.sqrt(3)/2, 
        -math.sqrt(2)/2, 
        -0.5,
        0, 
        0.5, 
        math.sqrt(2)/2, 
        math.sqrt(3)/2, 
        1
    ],
    ticktext=[
        '-1', 
        '-√3/2', 
        '-√2/2',
        '-0.5',
        '0', 
        '0.5', 
        '√2/2',
        '√3/2', 
        '1'
    ],
    showgrid=False,
)

fig.update_xaxes(
    range=[0, 360],
    tickmode='array',
    tickvals=list(range(0, 361, 30)),
    ticktext=[str(value) for value in range(0, 361, 30)],
    showgrid=True,
)

hline_args = dict(
    line_width=2,
    line_dash='solid', 
    line_color='#808080'
)
fig.add_hline(y=-1, **hline_args)
fig.add_hline(y=-math.sqrt(3)/2, **hline_args)
fig.add_hline(y=-math.sqrt(2)/2, **hline_args)
fig.add_hline(y=-0.5, **hline_args)
fig.add_hline(y=0, **hline_args)
fig.add_hline(y=0.5, **hline_args)
fig.add_hline(y=math.sqrt(2)/2, **hline_args)
fig.add_hline(y=math.sqrt(3)/2, **hline_args)
fig.add_hline(y=1, **hline_args)

fig.update_layout(hovermode='x unified')

fig.show()
```
